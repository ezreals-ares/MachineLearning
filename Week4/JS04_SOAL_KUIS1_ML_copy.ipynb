# Bagian 1 - Data Loading dan Data Imputation

## Soal 1 (5 poin)
1.   Lakukan inspeksi profile data
2.   **Variabel apa** yang memiliki **nilai yang hilang** (missing value) dan **berapa** jumlahnya?



```python
# Jawab Soal 1 - Inspeksi Missing Values
print('Info kolom dan missing values:')
print(df.isnull().sum())
print(f'\nTotal missing values: {df.isnull().sum().sum()}')
```

## Soal 2 (5 poin)
1. Lakukan proses data imputation pada fitur yang memiliki data yang hilang
2. Cek kembali apakah masih terdapat data yang hilang

```python
# Jawab Soal 2 - Data Imputation
# Isi missing values kolom string dengan modus
for col in df.select_dtypes(include='str').columns:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

# Cek apakah masih ada missing values
print('Missing values setelah imputasi:')
print(df.isnull().sum())
```

## Soal 3 (10 poin)
Inspeksi semua fitur kualitatif. Jika terdapat value yang **tidak sesuai**, **ganti dengan 'Others'** atau yang sesuai atau jika terdapat duplikasi karena **kesalahan penulisan**, lakukan penyesuaian.

```python
# Jawab Soal 3 - Inspeksi dan Perbaiki Fitur Kualitatif
cat_cols = df.select_dtypes(include='str').columns

for col in cat_cols:
    print(f'\n{col}: {df[col].unique()}')

# Bersihkan whitespace yang mungkin menyebabkan duplikasi
for col in cat_cols:
    df[col] = df[col].str.strip()

# Untuk kolom income, normalisasi nilai yang mirip (>50K. dan >50K sama)
df['income'] = df['income'].str.replace('.', '', regex=False)

print('\nNilai unik income setelah normalisasi:', df['income'].unique())
```

## Soal 2 - Analisis Visual (15 poin)
1. Fenomena apa yang terjadi pada distribusi data 'age'?
2. Jika terdapat data yang hilang pada variabel 'age', strategi apa yang Anda terapkan? Mengapa?
3. Berapa jumlah outlier pada setiap kategori 'income' berkaitan dengan 'hour-per-week'? Kategori apa yang paling banyak memiliki outlier?

```python
# Jawab Soal 2 - Analisis Visual

'''
1. Distribusi data 'age' bersifat right-skewed (condong ke kanan).
   Mayoritas penduduk berusia sekitar 25-45 tahun, dengan ekor panjang ke kanan (usia tua).

2. Jika terdapat missing value pada 'age', strategi yang tepat adalah menggunakan MEDIAN,
   karena data berdistribusi skewed (tidak normal). Median lebih robust terhadap outlier
   dan skewness dibanding mean.

3. Jumlah outlier dihitung dengan aturan IQR (Q1 - 1.5*IQR atau Q3 + 1.5*IQR).
   Berdasarkan boxplot, kategori <=50K cenderung memiliki lebih banyak outlier pada
   hours-per-week, namun kategori >50K juga menunjukkan outlier di sisi atas.
'''
# Hitung jumlah outlier per kategori income
for cat in df['income'].unique():
    subset = df[df['income'] == cat]['hours-per-week']
    Q1 = subset.quantile(0.25)
    Q3 = subset.quantile(0.75)
    IQR = Q3 - Q1
    n_outliers = ((subset < Q1 - 1.5*IQR) | (subset > Q3 + 1.5*IQR)).sum()
    print(f'Income {cat}: {n_outliers} outlier')
```

Pada bagian ini, Anda diminta untuk melakukan proses EDA dan pra pengolahan data sederhana pada dataset MNIST. Dataset MNIST merupakan data citra tulisan tangan untuk digil 0 hingga 9. Sebelum melakukan proses pengolahan, Anda akan dibantu dengan proses loading data dan inspeksi data.

```python
# Fetch data and inspect data shape
# Ganti tensorflow dengan sklearn karena tensorflow tidak tersedia
from sklearn.datasets import fetch_openml
import numpy as np
import matplotlib.pyplot as plt

# Load MNIST via sklearn
mnist_data = fetch_openml('mnist_784', version=1, as_frame=False, parser='auto')
X_all, y_all = mnist_data.data, mnist_data.target.astype(int)

# Reshape ke (n, 28, 28)
X_all = X_all.reshape(-1, 28, 28).astype(np.uint8)

# Split: 60000 train, 10000 test (sesuai MNIST standar)
X_train, X_test = X_all[:60000], X_all[60000:]
y_train, y_test = y_all[:60000], y_all[60000:]

print('Train shape:', X_train.shape)
print('Test shape:', X_test.shape)
```

```python
# Inspeksi Visual
plt.figure(figsize=(5,5))
for i in range(9):
    plt.subplot(3,3,i+1)
    plt.imshow(X_train[i], cmap="gray")
    plt.title(f"Label: {y_train[i]}")
    plt.axis("off")
plt.show()
```

## Soal 1 (10 poin)
1. Lakukan proses **upsampling** citra menjadi ukuran 32x32
2. Tampilakan 5 data hasil proses **upsampling**

Hint: Anda harus membuat array kosong untuk menampung hasil upsampling. Replace pada array X_test tidak dapat dilakukan karena data disimpan dalam bentuk ndarray yang memiliki ukuran fix (10000, (28,28))

```python
# Jawab Soal 1 - Upsampling citra MNIST dari 28x28 ke 32x32
from PIL import Image

# Buat array kosong untuk menampung hasil upsampling
X_test_32 = np.zeros((X_test.shape[0], 32, 32), dtype=np.uint8)

for i in range(len(X_test)):
    img = Image.fromarray(X_test[i])
    img_resized = img.resize((32, 32), Image.BILINEAR)
    X_test_32[i] = np.array(img_resized)

print('Shape hasil upsampling:', X_test_32.shape)

# Tampilkan 5 data hasil upsampling
plt.figure(figsize=(12, 3))
for i in range(5):
    plt.subplot(1, 5, i+1)
    plt.imshow(X_test_32[i], cmap='gray')
    plt.title(f'Label: {y_test[i]}')
    plt.axis('off')
plt.suptitle('5 Data Hasil Upsampling (32x32)')
plt.show()
```

## Soal 2 (10 poin)
Lakukan normalisasi nilai citra tiap piksel menjadi rentang 0-1

```python
# Jawab Soal 2 - Normalisasi pixel ke rentang 0-1
X_test_norm = X_test_32 / 255.0

print('Min pixel:', X_test_norm.min())
print('Max pixel:', X_test_norm.max())
print('Shape:', X_test_norm.shape)
```

## Soal 3 (10 poin)
Ubah metriks citra menjadi array 1 dimensi. Lakukan pada semua data test yang sudah di resize dan normalisasi.

Hint: Anda harus membuat holder array kosong untuk menampung hasilnya.

```python
# Jawab Soal 3 - Flatten citra 32x32 menjadi array 1 dimensi (1024)
X_test_flat = np.zeros((X_test_norm.shape[0], 32*32))

for i in range(len(X_test_norm)):
    X_test_flat[i] = X_test_norm[i].flatten()

print('Shape hasil flatten:', X_test_flat.shape)
```
