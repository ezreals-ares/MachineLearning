```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
dpath = '/home/ezreals_/Code/NoteBook/MachineLearning/Week2/Dataset/wbc.csv'
df = pd.read_csv(dpath)
df.head()
```

```python
df.info()
```

```python
df2 = df.copy()
df2 = df2.drop(columns='id')
df2.info()
```

```python
le = LabelEncoder()
df2['diagnosis'] = le.fit_transform(df2['diagnosis'])
df2.head(100)
```

```python
#Drop colum yang semuanya NaN
df2 = df2.dropna(axis=1, how='all')
#Pilih kolom numerik dan drop diagnosis
kolom_numerik = df2.drop(columns='diagnosis').select_dtypes(include=['float64', 'int64']).columns

std = StandardScaler()
df2[kolom_numerik] = std.fit_transform(df2[kolom_numerik])
df2.head()
```
